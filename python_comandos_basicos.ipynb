## 2. Variáveis e tipos

```python
inteiro = 10          # int
decimal = 3.14        # float
texto = "python"      # str
booleano = True       # bool
nada = None           # NoneType

print(type(inteiro), type(decimal), type(texto), type(booleano), type(nada))

# Conversão de tipos
print(int("42"), float("2.5"), str(100), bool(0), bool("oi"))

# Atribuição múltipla
a, b = 1, 2
a, b = b, a           # troca de valores
print(a, b)
```

## 3. Operadores

```python
x, y = 17, 5
print(x + y, x - y, x * y)   # soma, subtração, multiplicação
print(x / y)                 # divisão real
print(x // y)                # divisão inteira
print(x % y)                 # resto
print(x ** y)                # potência

# Comparação
print(x == y, x != y, x > y, x <= y)

# Lógicos
print(x > 10 and y < 10, x > 10 or y > 10, not (x > 10))

# Atribuição composta
x += 3; x -= 1; x *= 2
print(x)
```

## 4. Strings

```python
s = "  Python é Legal  "
print(len(s))
print(s.strip())                  # remove espaços das pontas
print(s.upper(), s.lower(), s.title())
print(s.replace("Legal", "Incrível"))
print(s.split())                  # lista de palavras
print("-".join(["a", "b", "c"]))  # junta com separador
print("Python" in s)              # contém?
print(s.strip().startswith("Py"), s.strip().endswith("l"))

# Indexação e fatiamento
t = "abcdef"
print(t[0], t[-1])      # primeiro e último
print(t[1:4])           # 'bcd'
print(t[::2])           # de 2 em 2
print(t[::-1])          # invertida
```

## 5. Listas (mutáveis)

```python
nums = [5, 3, 8, 1]
nums.append(10)          # adiciona no fim
nums.insert(0, 99)       # insere na posição
nums.remove(3)           # remove pelo valor
ultimo = nums.pop()      # remove e retorna o último
print(nums, ultimo)

nums.sort()              # ordena no lugar
print(nums)
print(sorted(nums, reverse=True))   # nova lista ordenada

print(len(nums), min(nums), max(nums), sum(nums))
print(nums[0], nums[-1], nums[1:3])
print(8 in nums, nums.index(8), nums.count(8))

lista_a = [1, 2]
lista_b = [3, 4]
print(lista_a + lista_b)   # concatenação
```

## 7. Dicionários (chave → valor)

```python
aluno = {"nome": "Ana", "idade": 20, "curso": "Computação"}

print(aluno["nome"])
print(aluno.get("email", "não informado"))   # valor padrão se não existir

aluno["idade"] = 21          # altera
aluno["email"] = "ana@x.com" # adiciona
del aluno["curso"]           # remove

print(aluno.keys())
print(aluno.values())
print(aluno.items())

for chave, valor in aluno.items():
    print(chave, "->", valor)

print("nome" in aluno)
```

## 8. Conjuntos (sem repetição)

```python
a = {1, 2, 3, 3, 2}
b = {3, 4, 5}
print(a)                 # {1, 2, 3}
print(a | b)             # união
print(a & b)             # interseção
print(a - b)             # diferença
a.add(10)
print(list(set([1, 1, 2, 2, 3])))   # remover duplicatas de lista
```

## 9. Condicionais

```python
nota = 7.5

if nota >= 7:
    print("Aprovado")
elif nota >= 5:
    print("Recuperação")
else:
    print("Reprovado")

# Operador ternário
status = "Aprovado" if nota >= 7 else "Não aprovado"
print(status)
```

## 10. Laços de repetição

```python
# for com range
for i in range(5):            # 0..4
    print(i, end=" ")
print()

for i in range(2, 10, 2):     # início, fim (exclusivo), passo
    print(i, end=" ")
print()

# for sobre lista, com índice
frutas = ["maçã", "banana", "uva"]
for i, fruta in enumerate(frutas):
    print(i, fruta)

# zip: percorre duas listas juntas
precos = [3.5, 2.0, 8.0]
for fruta, preco in zip(frutas, precos):
    print(fruta, preco)

# while
contador = 0
while contador < 3:
    print("contador:", contador)
    contador += 1

# break e continue
for n in range(10):
    if n == 2:
        continue      # pula este
    if n == 5:
        break         # encerra o laço
    print(n, end=" ")
```

## 11. Funções

```python
def saudacao(nome, saudacao="Olá"):
    """Retorna uma saudação personalizada."""
    return f"{saudacao}, {nome}!"

print(saudacao("Pietro"))
print(saudacao("Pietro", saudacao="Bom dia"))

def estatisticas(valores):
    return min(valores), max(valores), sum(valores) / len(valores)

menor, maior, media = estatisticas([4, 8, 15, 16, 23, 42])
print(menor, maior, media)

# *args e **kwargs
def soma(*numeros):
    return sum(numeros)
print(soma(1, 2, 3, 4))

# Função lambda
quadrado = lambda x: x ** 2
print(quadrado(6))

# map e filter
print(list(map(quadrado, [1, 2, 3])))
print(list(filter(lambda x: x % 2 == 0, range(10))))
```

## 12. List comprehension

```python
quadrados = [x**2 for x in range(1, 6)]
pares = [x for x in range(20) if x % 2 == 0]
dobro_dict = {x: x * 2 for x in range(4)}
print(quadrados)
print(pares)
print(dobro_dict)
```

## 13. Arquivos

```python
# Escrever
with open("exemplo.txt", "w", encoding="utf-8") as f:
    f.write("linha 1\n")
    f.write("linha 2\n")

# Ler tudo
with open("exemplo.txt", "r", encoding="utf-8") as f:
    conteudo = f.read()
print(conteudo)

# Ler linha a linha
with open("exemplo.txt", "r", encoding="utf-8") as f:
    for linha in f:
        print(linha.strip())

# Acrescentar ao final: modo "a"  |  Ler: "r"  |  Escrever (apaga): "w"
```

## 14. Tratamento de exceções

```python
try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("Não é possível dividir por zero")
except Exception as e:
    print("Outro erro:", e)
else:
    print("Deu certo:", resultado)
finally:
    print("Sempre executa")

# Levantar um erro manualmente
def raiz(x):
    if x < 0:
        raise ValueError("x não pode ser negativo")
    return x ** 0.5

print(raiz(9))
```

## 16. Classes e objetos

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def apresentar(self):
        return f"Sou {self.nome} e tenho {self.idade} anos"

    def __str__(self):
        return f"Pessoa({self.nome}, {self.idade})"

p = Pessoa("Carlos", 30)
print(p.apresentar())
print(p)

# Herança
class Aluno(Pessoa):
    def __init__(self, nome, idade, curso):
        super().__init__(nome, idade)
        self.curso = curso

    def apresentar(self):
        return super().apresentar() + f" e estudo {self.curso}"

print(Aluno("Bia", 19, "Ciência da Computação").apresentar())
```
